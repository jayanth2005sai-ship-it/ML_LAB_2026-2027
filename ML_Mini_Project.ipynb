```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
wine = load_wine()

dataset = pd.DataFrame(wine.data, columns=wine.feature_names)
dataset["wine_class"] = wine.target

display(dataset.head())

print("Dataset shape:", dataset.shape)
print("Missing values:", dataset.isnull().sum().sum())
print("\nSamples per class:")
print(dataset["wine_class"].value_counts().sort_index())
```

```python
X = dataset.drop("wine_class", axis=1)
Y = dataset["wine_class"]

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.3,
    random_state=2,
    stratify=Y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=2
    ),
    "Logistic Regression": LogisticRegression(
        max_iter=1000
    ),
    "k-Nearest Neighbors": KNeighborsClassifier(
        n_neighbors=5
    ),
    "Support Vector Machine": SVC(
        kernel="rbf"
    )
}

for name, model in models.items():
    model.fit(X_train_scaled, Y_train)
    print(f"{name} trained successfully")
```

```python
X = dataset.drop("wine_class", axis=1)
Y = dataset["wine_class"]

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.3,
    random_state=2,
    stratify=Y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
predictions = {}
results = []

print("       --- Analysis Report of All Algorithms ---\n")

for name, model in models.items():
    y_pred = model.predict(X_test_scaled)
    predictions[name] = y_pred

    accuracy = accuracy_score(Y_test, y_pred)

    results.append({
        "Algorithm": name,
        "Accuracy (%)": accuracy * 100
    })

    print(f"\n{name}")
    print("-" * 60)

    print(classification_report(
        Y_test,
        y_pred,
        target_names=wine.target_names,
        zero_division=0
    ))

    print(f"Accuracy: {accuracy * 100:.2f}%")
```
